##### Data Pull
Find the nearest coordinates containing all the required datasets and its index value

Upload CSV file containing: "location_name, latitude, longitude, file_path"

```python
import pandas as pd
import numpy as np
import h5pyd
from pathlib import Path

locations = pd.read_csv("ATON_Locations.csv")
display(locations)

# location = locations.iloc[0]
# name = location["location_name"]
# lat = location["latitude"]
# lon = location["longitude"]
# file_path = location["file_name"]

parameters = [
    "significant_wave_height","peak_period","energy_period",
    "omni-directional_wave_power","mean_wave_direction","spectral_width",
    "directionality_coefficient"
]

names = []
lats = []
lons = []
file_paths = []
nearest_lat = []
nearest_lon = []
index = []
```

Loop through the files

```python
for _, location in locations.iterrows():
    name = location["location_name"]
    lat = location["latitude"]
    lon = location["longitude"]
    file_path = location["file_name"]
    names.append(name)
    lats.append(lat)
    lons.append(lon)
    file_paths.append(file_path)

```

Run through each of the file paths and save the nearest latitudes, longitudes, and indexes in arrays



```python
for i, location in locations.iterrows():

    path = file_paths[i]
    path_lat = lats[i]
    path_lon = lons[i]
    path_name = names[i]

    with h5pyd.File(path, "r"
    
    ) as f:

        coords = f["coordinates"][:]

        # Approximate geographic distance
        distance = (
            (coords[:, 0] - path_lat)**2
            + (np.cos(np.radians(path_lat)) *
            (coords[:, 1] - path_lon))**2
        )

        idx = np.argmin(distance)

        nearest_lat.append(coords[idx][0])
        nearest_lon.append(coords[idx][1])
        index.append(idx)

        print("Location:", path_name)
        print("Nearest grid point index:", idx)
        print("Nearest coordinates:", coords[idx])
index = np.array(index)
nearest_lat = np.array(nearest_lat)
nearest_lon = np.array(nearest_lon)
```

Add on to the existing locations csv and save as a new csv called "nearest_locations"

```python
locations['Grid_Index'] = index
locations['Nearest_Latitude'] = nearest_lat
locations['Nearest_Longitude'] = nearest_lon
locations.to_csv(r"C:\Users\gupta\projects\mhkit-repo\notebooks\nearest_locations.csv", index=False)

print(locations)
```
